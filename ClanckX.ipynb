```python
TICKER = "SPY"
START_CASH = 10000

df = yf.download(TICKER, start="2020-01-01")
close = df["Close"].squeeze()

sma20 = close.rolling(20).mean()
sma50 = close.rolling(50).mean()
position = (sma20 > sma50).astype(int).shift(1).fillna(0)

daily_return = close.pct_change().fillna(0)
strategy_return = daily_return * position

strategy_value = START_CASH * (1 + strategy_return).cumprod()
hold_value = START_CASH * (1 + daily_return).cumprod()

print(f"Ticker: {TICKER}")
print(f"Start cash: ${START_CASH:,.0f}")
print(f"ClanckX Strategy end: ${strategy_value.iloc[-1]:,.0f}")
print(f"Buy & Hold end: ${hold_value.iloc[-1]:,.0f}")
```

```python
import requests, time
from IPython.display import clear_output

GREEN = "\033[92m"
RED = "\033[91m"
WHITE = "\033[0m"
RESET = "\033[0m"

BASES = [
    "https://api.binance.com",
    "https://data-api.binance.vision",
    "https://api.binance.us",
]

def get_data(path):
    for base in BASES:
        try:
            data = requests.get(base + path, timeout=5).json()
            if 'price' in data or isinstance(data, list):
                return data
        except:
            continue
    return None

def get_price(symbol):
    data = get_data(f"/api/v3/ticker/price?symbol={symbol}")
    return float(data['price']) if data else 0

coins = [
    ("SOL","SOLUSDT"),
    ("BTC","BTCUSDT"),
    ("DOGE","DOGEUSDT"),
    ("ETH","ETHUSDT"),
    ("XRP","XRPUSDT"),
    ("BNB","BNBUSDT"),
    ("ADA","ADAUSDT"),
    ("PEPE","PEPEUSDT"),
    ("SHIB","SHIBUSDT"),
    ("TRX","TRXUSDT"),
]

old = {name: 0 for name, sym in coins}

while True:
    clear_output(wait=True)
    print("=== CLANCKX LIVE - 10 COINS - WORLDWIDE ===")
    print("Nigeria, Ukraine, USA, UK - Everywhere")
    print("")
    for name, symbol in coins:
        new_price = get_price(symbol)
        old_price = old[name]

        if new_price > old_price:
            color = GREEN
            arrow = "UP ↑"
        elif new_price < old_price:
            color = RED
            arrow = "DOWN ↓"
        else:
            color = WHITE
            arrow = "SAME -"

        print(f"{name}: {color}{new_price} {arrow}{RESET} | Old: {old_price}")
        old[name] = new_price

    print("")
    print("Checking every 5 seconds... Worldwide")
    time.sleep(5)
```
